```python
times = ['Real Madrid', 'Barcelona', 'PSG', 'Chelsea', 'Bayern']
print('Os três primeiros colocados:', ', '.join(times[:3]))
print('Os dois ultimos colocados:', ', '.join(times[3:]))
print('Os times em ordem alfabetica:', ', '.join(sorted(times)))
print('O Barcelona se encontra na posição:', times.index('Barcelona'))
```

**Questão 4**

```python
pessoas = []
for i in range(3):
    nome = input(f"Nome da pessoa {i+1}: ")
    peso = float(input(f"Peso da pessoa {i+1}: "))
    pessoas.append({"nome": nome, "peso": peso})

mais_pesada = max(pessoas, key=lambda p: p["peso"])
mais_leve = min(pessoas, key=lambda p: p["peso"])

print(f"Pessoa mais pesada: {mais_pesada['nome']} ({mais_pesada['peso']} kg)")
print(f"Pessoa mais leve: {mais_leve['nome']} ({mais_leve['peso']} kg)")
```

**Questão 5**

```python
n = int(input("Quantas pessoas? "))
pessoas = []
for i in range(n):
    nome = input(f"Nome da pessoa {i+1}: ")
    idade = int(input(f"Idade da pessoa {i+1}: "))
    sexo = input(f"Sexo da pessoa {i+1} (M/F): ")
    pessoas.append({"nome": nome, "idade": idade, "sexo": sexo})

media_idade = sum(p["idade"] for p in pessoas) / n
mulheres_menores_20 = sum(1 for p in pessoas if p["sexo"].upper() == "F" and p["idade"] < 20)

print(f"Média de idade do grupo: {media_idade:.2f}")
print(f"Mulheres com menos de 20 anos: {mulheres_menores_20}")
```

**Questão 8**

```python
produtos = []
for i in range(3):
    nome = input(f"Nome do produto {i+1}: ")
    preco = float(input(f"Preço do produto {i+1}: "))
    quantidade = int(input(f"Quantidade em estoque do produto {i+1}: "))
    produtos.append({"nome": nome, "preco": preco, "quantidade": quantidade})

for produto in produtos:
    valor_total = produto["preco"] * produto["quantidade"]
    print(f"{produto['nome']}: R$ {valor_total:.2f}")
```
