**Questão** 1

```python
nome = input("Digite seu nome completo: ")

print("Nome em maiúsculas:", nome.upper())
print("Nome em minúsculas:", nome.lower())

print("Quantidade de letras no nome:", len(nome.replace(" ","")))

partes_nome = nome.split()
novo_nome = " ".join(partes_nome[:-1] + ["do", "Inatel"])
print("Nome com o último sobrenome trocado:", novo_nome)
```

**Questão 2**

```python
numero = int(input("Digite o número para ver a tabuada: "))
inicio = int(input("Digite o início do intervalo: "))
fim = int(input("Digite o fim do intervalo: "))

print(f"\nTabuada do {numero} de {inicio} até {fim}:")
for i in range(inicio, fim + 1):
    print(f"{numero} x {i} = {numero * i}")
```

**Questão 3**

```python
sexo = input("Digite seu sexo (M/F): ").upper()

while sexo != "M" and sexo != "F":
    print("Valor inválido!")
    sexo = input("Digite seu sexo (M/F): ").upper()

if sexo == "M":
    print("Você é homem.")
else:
    print("Você é mulher.")
```

**Questão 4**

```python
distancia = float(input('Digite a distância em km: '))

if distancia <= 200:
    valor = distancia * 0.50
else:
    valor = distancia * 0.45

print(f"Valor da passagem: R$ {valor:.2f}")
```

**Questão 5**

```python
numero = int(input("Digite um número entre 1000 e 9999: "))

while numero < 1000 or numero > 9999:
    print("Número fora do intervalo!")
    numero = int(input("Digite um número entre 1000 e 9999: "))

unidade = numero % 10
dezena = (numero // 10) % 10
centena = (numero // 100) % 10
milhar = numero // 1000

print("Unidade:", unidade)
print("Dezena:", dezena)
print("Centena:", centena)
print("Milhar:", milhar)
```

**Questão 6**

```python
import math

numero = float(input("Digite um número decimal: "))

print("Raiz quadrada:", math.sqrt(numero))
print("Função teto (ceil):", math.ceil(numero))
print("Função chão (floor):", math.floor(numero))
print("Parte inteira:", int(numero))
```

**Questão 7**

```python
palavra = input("Digite uma palavra: ")

vogais = "aeiouAEIOU"
contador_vogais = 0
tem_letra_a = False

for letra in palavra:
    print(letra.upper())
    if letra in vogais:
        contador_vogais += 1
    if letra.lower() == "a":
        tem_letra_a = True

print("\nQuantidade de vogais:", contador_vogais)
print("Tem letra a?", "Sim" if tem_letra_a else "Não")
```

**Questão 8**

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

adicao = num1 + num2
subtracao = num1 - num2
multiplicacao = num1 * num2
divisao = num1 / num2
resto = num1 % num2
potencia = num1 ** num2

print("Adição:", adicao)
print("Subtração:", subtracao)
print("Multiplicação:", multiplicacao)
print("Divisão:", divisao)
print("Resto da divisão:", resto)
print("Potência:", potencia)

if adicao % 2 == 0:
    print("O resultado da adição é PAR.")
else:
    print("O resultado da adição é ÍMPAR.")
```
