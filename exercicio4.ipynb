Questão 2

```python
campo = np.zeros((2, 2), dtype=int)
linha_bomba = np.random.randint(0, 2)
coluna_bomba = np.random.randint(0, 2)
campo[linha_bomba, coluna_bomba] = 1

total_posicoes_seguras = campo.size - 1
posicoes_acertadas = set()
jogadas = 0
jogo_ativo = True

print("Escolha uma posição (0 ou 1).\n")

while jogo_ativo:
    linha = int(input("Linha (0 ou 1): "))
    coluna = int(input("Coluna (0 ou 1): "))
    jogadas += 1

    if (linha, coluna) == (linha_bomba, coluna_bomba):
        if jogadas <= 3:
            print("Game Over! :( Try Again!")
        jogo_ativo = False
    else:
        posicoes_acertadas.add((linha, coluna))
        print(f"Posição segura! ({len(posicoes_acertadas)}/{total_posicoes_seguras})\n")
        if len(posicoes_acertadas) == total_posicoes_seguras:
            print("Congratulations! You beat the game! :)")
            jogo_ativo = False
```

Questão 4

```python
np.random.seed(10)
matriz4x4 = np.random.randint(1, 51, size=(4, 4))

media_linhas = matriz4x4.mean(axis=1)
media_colunas = matriz4x4.mean(axis=0)
print("Média de cada linha:", media_linhas)
print("Média de cada coluna:", media_colunas)

maior_media_linha = media_linhas.max()
maior_media_coluna = media_colunas.max()
print(f"Maior média entre as linhas: {maior_media_linha}")
print(f"Maior média entre as colunas: {maior_media_coluna}")

valores, contagens = np.unique(matriz4x4, return_counts=True)
print("\nQuantidade de aparições de cada número gerado:")
for valor, contagem in zip(valores, contagens):
    print(f"  Número {valor}: {contagem} vez(es)")

numeros_duas_vezes = valores[contagens == 2]
print("\nNúmeros que aparecem 2 vezes:", numeros_duas_vezes)
```

Questão 5

```python
zeros_3x3 = np.zeros((3, 3))
uns_3x3 = np.ones((3, 3))

soma_matrizes = zeros_3x3 + uns_3x3
print("Soma:")
print(soma_matrizes)

escalar = 5
resultado_escalar = soma_matrizes * escalar
print(f"\nResultado multiplicado pelo escalar {escalar}:")
print(resultado_escalar)

vetor_final = resultado_escalar.reshape(9)
print("\nArray unidimensional de 9 elementos:")
print(vetor_final)
```

Questão 6

```python
array_arange = np.arange(10, 31, 2)

maior_valor = array_arange.max()
indice_maior = array_arange.argmax()
menor_valor = array_arange.min()
indice_menor = array_arange.argmin()
soma_total = array_arange.sum()

print(f"Maior valor: {maior_valor} (índice {indice_maior})")
print(f"Menor valor: {menor_valor} (índice {indice_menor})")
print(f"Soma: {soma_total}")
```
