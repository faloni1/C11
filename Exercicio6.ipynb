### Questão 1

```python
import pandas as pd

seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})
```

### Questão 2

```python
print(seriesAno1.sum())
print(seriesAno2.sum())
```

### Questão 3

```python
crescimento = seriesAno2 - seriesAno1
crescimento
```

### Questão 4

```python
crescimento[crescimento > 0]
```

### Questão 5

```python
futuro = seriesAno2 + 2 * crescimento
futuro.nlargest(1)
```
