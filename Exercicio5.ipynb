```python
import numpy as np

dados = np.genfromtxt('space.csv', delimiter=';', names=True, dtype=None, encoding='utf-8-sig', autostrip=True)
```

### Questão 1

```python
status = dados['Status_Mission']
sucesso = np.sum(status == 'Success')
porcentagem_sucesso = (sucesso / len(status)) * 100
print(f'Porcentagem de missões com sucesso: {porcentagem_sucesso:.2f}%')
```

### Questão 2

```python
custos = dados['Cost']
custos_validos = custos[custos > 0]
media_custos = np.mean(custos_validos)
print(f'Média de gastos das missões: {media_custos:.2f} milhões de dólares')
```

### Questão 3

```python
localizacoes = dados['Location']
missoes_eua = np.sum(np.char.find(localizacoes, 'USA') != -1)
print(f'Missões realizadas pelos Estados Unidos: {missoes_eua}')
```

### Questão 4

```python
spacex = dados[dados['Company_Name'] == 'SpaceX']
custos_spacex = spacex['Cost']
missoes_spacex = spacex['Detail']
indice_max = np.argmax(custos_spacex)
print(f'Missão mais cara da SpaceX: {missoes_spacex[indice_max]}')
print(f'Custo: {custos_spacex[indice_max]} milhões de dólares')
```

### Questão 5

```python
empresas, quantidades = np.unique(dados['Company_Name'], return_counts=True)
for empresa, quantidade in zip(empresas, quantidades):
    print(f'{empresa}: {quantidade}')
```

### Questão 6

```python
status_foguete = dados['Status_Rocket']
retirados = np.sum(status_foguete == 'StatusRetired')
porcentagem_retirados = (retirados / len(status_foguete)) * 100
print(f'Porcentagem de foguetes com status StatusRetired: {porcentagem_retirados:.2f}%')
```
